# Array Copy vs View

```python
a=np.array([10,20,30,40])
b=a.copy()
b[0]=100
print(a)
print(b)
```

```python
a=np.array([10,20,30,40])
b=a.view()
b[0]=100
print(a)
print(b)
```

# base attribute:-

```python
import numpy as np
a = np.array([1, 2, 3, 4])
b = a.copy()
c = a.view()
print(b.base)
print(c.base)
```

### Array Mathematics

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print(a + b)
print(a - b)
print(a * 2)
print(a / 2)
print(a ** 2)
```

### Universal functions (ufunc)

```python
a=np.array([1,4,9])
print(np.sqrt(a))
print(np.exp(a))
print(np.log(a))
print(np.sin(a))
print(np.cos(a))
print(np.abs(a))
```

### Aggregation Function 

```python
a=np.array([3,7,2,9,4])
print(np.sum(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.min(a))
print(np.max(a))
print(np.argmin(a))
print(np.argmax(a))
```
