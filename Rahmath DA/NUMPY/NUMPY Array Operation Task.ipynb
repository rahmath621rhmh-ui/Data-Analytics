```python
# 4. Given the array:
# a = np.array([[1, 2], [3, 4]])
# Create a view of the array, change the first element of the view to 100 , and
# print both the original array and the view. Check the base attribute.
```

```python
a=np.array([[1,2],[3,4]])
b=a.view()
b[0,0]=100
print(a)
print(b)
```

```python
print(a.base)
print(b.base)
```
