```python
#  Given the following NumPy array, perform the
# operations below.
```

```python
# a. Display all elements greater than the overall mean of the array.
```

```python
arr = np.array([[12, 5, 18, 7],
               [25, 10, 3, 20],
               [8, 30, 15, 2]])
```

```python
a=np.array([[12,5,18,7],
              [25,19,3,20],
              [8,30,15,22]])
print(a.mean())
print(a[a>5])
```

```python
# b. Find the positions (row and column indices) of all elements divisible by 5.
```

```python
print(np.where(a%5==0))
```

```python
print(np.argwhere(a%5==0))
```

```python
# c. Replace all elements less than 10 with -1 without using a loop.
```

```python
a=np.where(a<10,-1,a)
print(a)
a[a>10] =-1
print(a)
```

```python
# d. Calculate the sum of each row and identify the row with the highest sum.
```

```python
a=(np.sum(arr,axis=1))
print(a)
print(np.max(a))
print(np.argmax(a))
```

```python
# e. Sort each row in ascending order and display the result.
```

```python
print(np.sort(arr,axis=1))
```

```python
# f. Extract the last two columns and reshape them into a one-dimensional array.
```

```python
a=arr[:,-2:]
print(a)
print(a.reshape(6))
```

```python
# g. Create a flattened copy and a raveled view of the array. Modify one element
# in each and observe the effect on the original array.
```

```python
flat_arr = arr.flatten()
flat_arr[0] = 100
print(arr)
print(flat_arr)
ravel_arr = arr.ravel()
ravel_arr[1] = 200
print(arr)
print(ravel_arr)
```

```python
# h. Split the flattened array into three equal parts and display each part.
```

```python
flat = arr.flatten()
parts = np.split(flat, 3)
for part in parts:
 print(part)
```

```python
# i. Calculate the square root of all elements in the original array and round the
# results to two decimal places.
```

```python
print(np.round(np.sqrt(arr), 2))
```

```python
# j. Find the minimum and maximum values and their corresponding indices in the
# original array.
```

```python
print("Minimum value:", np.min(arr))
print("Minimum index:", np.argmin(arr))
print("Maximum value:", np.max(arr))
print("Maximum index:", np.argmax(arr))
```
