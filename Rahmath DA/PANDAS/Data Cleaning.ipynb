<h2 style="color:#1F4E78; margin:0;">
DATA INSPECTION AND CLEANING
</h2>

</div>

In this section, the dataset is inspected and cleaned by checking missing values, duplicate records,
inconsistent values and incorrect data types

```python
 empty_count = df.apply( 
     lambda col: col.astype('string').str.fullmatch(r'\s*').fillna(False).sum() 
 ) 
  
 print(empty_count)
```

```python
import numpy as np
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)
```

```python
df.drop("Unnamed: 8",axis=1,inplace=True)
```

```python
df
```

```python
df["Score"]=pd.to_numeric(df["Score"],errors="coerce")
```

```python
df[" Votes "]=pd.to_numeric(df[" Votes "],errors="coerce")
```

```python
df["Duration"]=pd.to_numeric(df["Duration"],errors="coerce")
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df.duplicated().sum()
```

```python
df["IMBD title ID"]=df["IMBD title ID"].fillna(df["IMBD title ID"].mode()[0],inplace=True)
```

```python
df["Original titlÊ"]=df["Original titlÊ"].fillna(df["Original titlÊ"].mode()[0],inplace=True)
```

```python
df["Release year"]=df["Release year"].fillna(df["Release year"].mode()[0],inplace=True)
```

```python
df["Genrë¨"]=df["Genrë¨"].fillna(df["Genrë¨"].mode()[0],inplace=True)
```

```python
df["Duration"]=df["Duration"].fillna(df["Duration"].mean(),inplace=True)
```

```python
df["Country"]=df["Country"].fillna(df["Country"].mode()[0],inplace=True)
```

```python
df["Content Rating"]=df["Content Rating"].fillna("unknown",inplace=True)
```

```python
df["Director"]=df["Director"].fillna(df["Director"].mode()[0],inplace=True)
```

```python
df["Income"]=df["Income"].fillna(df["Income"].mode()[0],inplace=True)
```

```python
df[" Votes "]=df[" Votes "].fillna(df[" Votes "].median(),inplace=True)
```

```python
df["Score"]=df["Score"].fillna(df["Score"].mean(),inplace=True)
```
