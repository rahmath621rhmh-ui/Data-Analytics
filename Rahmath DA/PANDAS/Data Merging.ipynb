# <CENTER> Pandas Data Integration and Transformation

## MERGING DATA USING PANDAS

```python
import pandas as pd
```

### Basic Merging:

```python
df1=pd.DataFrame({"id":[1,2,3],
                  "name":["Asha","Binu","Cathy"]})

df2=pd.DataFrame({"id":[1,2,4],
                   "marks":[85,90,75]})

merged_df=pd.merge (df1, df2, on="id")
print(merged_df)
```

# <CENTER> Types of joins:

## Inner Join(default)

### only common rows:

```python
pd.merge(df1,df2, on="id", how="inner")
```

## Left Join

### All rows from left DataFrame and common from right:

```python
pd.merge(df1,df2, on="id", how="left")
```

## Right Join

### All rows from right DataFrame and common from left

```python
pd.merge(df1,df2, on="id", how="right")
```

## Outer Join

### All rows from both:

```python
pd.merge(df1,df2, on="id", how="outer")
```

## Merge on Different Column Names

```python
df1=pd.DataFrame({"id":[1,2,3],
                  "name":["Asha","Binu","Cathy"]})

df2=pd.DataFrame({"student_id":[1,2,4],
                  "marks":[85,90,75]})
merged_df=pd.merge(df1,df2,left_on="id",right_on="student_id")
print(merged_df)            
```

```python
merg1=pd.merge(df1,df2,left_on="id",right_on="student_id" ,how="left")
print(merg1)
```

```python
merg2=pd.merge(df1,df2,left_on="id",right_on="student_id" ,how="right")
print(merg2)             
```

```python
merg3=pd.merge(df1,df2,left_on="id",right_on="student_id" ,how="inner")
print(merg3)
```

```python
merg4=pd.merge(df1,df2,left_on="id",right_on="student_id" ,how="outer")
print(merg4)
```

## Concatenation

```python
pd.concat([df1,df2], axis=0)  (row)
```

```python
pd.concat([df1,df2], axis=1)
```

### Apply

```python
marks=pd.Series([45,60,75,90])
def grade(x):
    if x>=50:
        return "Pass"
    else:
        return "Fail"
print(marks.apply(grade))        
```

### Apply Using Lambda Function

```python
marks=pd.Series([45,60,75,90])
result=marks.apply(lambda x:x+5)
print(result)
```

### Apply()on DataFrame Columns

```python
df=pd.DataFrame({"maths":[80,70,90],
                 "science":[85,60,95]})
print(df.apply(max))
```
