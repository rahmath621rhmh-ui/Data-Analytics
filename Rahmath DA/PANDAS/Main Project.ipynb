<h2 style="color:#38761D; margin:0;">GROUPBY ANALYSIS</h2>
</div>
GroupBy is used to group data based on a specific column.

```python
df.groupby("Study Hours")["Grades"].mean()
```

**Insight:**
- At 0.8 study hours, average grade is 33.57.
- At 9.9 study hours, average grade is 91.

Finding: Higher study hours generally show higher average grades.

```python
df.groupby("Sleep Hours")["Grades"].mean()
```

**Insight: Average grades vary across different sleep-hour groups. No clear increasing pattern is observed.**

```python
df.groupby("Attendance (%)")["Grades"].mean()
```

**Insight:**
- At 40% attendance, average grade is 38.75.
- At 99% attendance, average grade is 67.

Finding: Higher attendance groups generally show higher average grades.

```python
df.groupby("SES_Score")["Grades"].mean()
```

**Insight: Most SES Score values are unique, making group comparisons difficult.**
