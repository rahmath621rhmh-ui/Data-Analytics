# <CENTER> Pandas DataFrame

```python
import pandas as pd
```

### Creating a DataFrame from a Dictionary

```python
data={"Name":["Anu","Rahul","Priya"],
      "Age":[21,22,20],
      "Mark":[85,90,88]}
df=pd.DataFrame(data)
print(df)
```

### Creating a DataFrame from a List

```python
data=[["Anu",21,85],
      ["Rahul",22,90],
       ["Priya",20,88]]
df=pd.DataFrame(data,columns=["Name","Age","Mark"])
print(df)
```

### Creating a DataFrame with a Custom Index

```python
data={"Name":["Anu","Rahul","Priya"],
      "Mark":[85,90,88]}
df=pd.DataFrame(data,index=["s1","s2","s3"])
print(df)
```

### Displaying Columns - df.columns

```python
print(df.columns)
```

### Selecting a Single Column

### Selecting a Multiple Column

### Selecting Rows Using loc

```python
data={"Name":["Anu","Rahul","Priya"],
      "Mark":[85,90,88]}
df=pd.DataFrame(data,index=["s1","s2","s3"])
print(df.loc["s2"])
```

### Basic Statistics - describe()


```python
df = pd.DataFrame({"Mark": [70, 80, 90, 85, 75]})
print(df.describe())
```

### Filtering Rows

```python
df= pd.DataFrame({"Name": ["Anu", "Rahul", "Priya", "Arun"],
                   "Mark": [85, 90, 45, 75]})
print(df[df["Mark"] > 80])
```

### Sorting a DataFrame

```python
# Ascending order
df = df.sort_values("Mark")
print(df)
# Descending order
df = df.sort_values("Mark", ascending=False)
print(df)
```

### Checking for Missing Values - isna()

```python
import numpy as np
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya"],
                   "Mark": [85, np.nan, 90]})
print(df.isna())
```

### Counting Missing Values

```python
print(df.isna().sum())
```

### Filling Missing Values - fillna()

```python
df["Mark"] = df["Mark"].fillna(0)
print(df)
```

### Removing Rows with Missing Values - dropna()

```python
df = df.dropna()
print(df)
```

### Viewing First and Last Rows - head() and tail()

```python
print(df.head()) # first 5 rows
print(df.head(3)) # first 3 rows
```

```python
print(df.tail()) # last 5 rows
print(df.tail(2)) # last 2 rows
```

### DataFrame Arithmetic

```python
df["Mark"] = df["Mark"] + 5 
# Similarly:
df["Mark"] * 2
df["Mark"] - 5
df["Mark"] / 2
```
