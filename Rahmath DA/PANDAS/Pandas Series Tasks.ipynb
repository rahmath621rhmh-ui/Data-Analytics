#  <CENTER> Pandas Series Task

```python
import pandas as pd
```

```python
# 1. Create a Pandas Series containing the values 10, 20, 30, 40, 50 .
```

```python
s=pd.Series([10,20,30,40,50])
print(s)
```

```python
# 2. Create a Pandas Series from the list:
# ["Apple", "Banana", "Mango", "Orange"] .
```

```python
s=pd.Series(["Apple","Banana","Mango","Orange"])
print(s)
```

```python
# 3. Create a Pandas Series with custom indexes A, B, C, D and values 100, 200,
# 300, 400 .
```

```python
s=pd.Series([100,200,300,400],index=["A","B","C","D"])
print(s)
```

```python
# 4. Create a Pandas Series and access the third element.
```

```python
s=pd.Series([10,20,30,40,50])
print(s.loc[3])
```

```python
# 5. Create a Pandas Series and display the first three elements
```

```python
s=pd.Series([10,20,30,40,50])
print(s.loc[:2])
```

```python
# 6. Create a Pandas Series containing five numbers and find its sum and mean.
```

```python
a=pd.Series([20,40,60,80,90])
print(a.sum())
print(a.mean())
```

```python
# 7. Create a Pandas Series containing student marks and find the maximum
# and minimum marks.
```

```python
mark=pd.Series([20,40,60,80,90])
print(mark.max())
print(mark.min())
```

```python
# 8. Create a Pandas Series containing numbers and display only the values
# greater than 50.
```

```python
a=pd.Series([40,45,55,60,65])
print(a[a>50])
```

```python
# 9. Create a Pandas Series containing duplicate values and display the unique
# values.
```

```python
a=pd.Series([50,30,50,60,70])
print(a.unique())
```

```python
# 10. Create a Pandas Series and display its index, values, data type, and
# number of elements.
```

```python
a=pd.Series([10,20,30,40,50])
print(a.index)
print(a.dtype)
print(a.size)
```

```python
# 11. Create a Pandas Series from the following dictionary:
# {"A": 100, "B": 200, "C": 300, "D": 400}
```

```python
data={
    "A":100,
    "B":200,
    "C":300,
    "D":400
}
s=pd.Series(data)
print(s)
```

```python
# 12. Create a Pandas Series with indexes A, B, C and values 10, 20, 30 . Add a
# new element with index D and value 40 .
```

```python
s=pd.Series([10,20,30],index=["A","B","C"])
s["D"]=40
print(s)
```

```python
# 13. Create a Pandas Series with indexes A, B, C, D and values 10, 20, 30, 40 .
# Remove the element with index C .
```

```python
s=pd.Series([10,20,30,40],index=["A","B","C","D"])
print(s.drop("C"))
```

```python
# 14. Create a Pandas Series containing [25, 10, 45, 30, 15] and sort the values
# in ascending order.
```

```python
s=pd.Series([25, 10, 45, 30, 15])
print(s.sort_values())
```

```python
# 15. Create a Pandas Series containing [10, 20, 30, 40, 50] and check whether
# the value 30 exists in the Series.
```

```python
a=pd.Series([10, 20, 30, 40, 50])
print(30 in a.values)
```
