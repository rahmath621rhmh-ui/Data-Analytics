```python
import pandas as pd
```

## Creating series from a list

```python
s=pd.Series([10,20,30,40,50])
print(s)
```

## Creating a Series from list of string

```python
s=pd.Series(["Apple","Banana","Mango","Orange"])
print(s)
```

## Series with a custom index

```python
s=pd.Series([100,200,300,400],
           index=["A","B","C","D"])
print(s)
```

 ## Creating a Series from a Dictionary

```python
data = {
 "A": 100,
 "B": 200,
 "C": 300,
 "D": 400
}
s = pd.Series(data)
print(s)
```

## Accessing Series Elements - iloc

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.iloc[2])
```

## loc - Label-Based Access

```python
s = pd.Series([100, 200, 300, 400],
 index=["A", "B", "C", "D"])
print(s.loc["C"])
```

## Slicing a Series

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.iloc[:3])
```

## First and Last Elements

```python
# First element
print(s.iloc[0])
# Last element
print(s.iloc[-1])
# First three elements
print(s.iloc[:3])
# Last two elements
print(s.iloc[-2:])
```

## Series Arithmetic Operations

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s + 10)
```

## Mathematical Functions

```python
marks = pd.Series([75, 88, 92, 65, 80])
print(marks.sum()) # Total of all values
print(marks.mean()) # Average value
print(marks.max()) # Largest value
print(marks.min()) # Smallest value
```

## Important Statistical Methods

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.sum())
print(s.mean())
print(s.median())
print(s.max())
print(s.min())
```

## Filtering a Series

```python
s = pd.Series([25, 60, 45, 80, 35, 70])
print(s[s > 50])
```

## Checking Whether a Value Exists

```python
s = pd.Series([10, 20, 30, 40, 50])
print(30 in s.values)
print(100 in s.values)
```
