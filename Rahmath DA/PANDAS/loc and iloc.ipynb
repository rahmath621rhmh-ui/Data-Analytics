## <CENTER> Data Selection Using loc and Iloc

```python
import pandas as pd
```

## 1. loc (Label-Based Indexing)

```python
data= {
"Name": ["Asha","Rahul","Anu"],
"Age": [20,21,19],
"City": ["Kochi","Delhi","Chennai"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.loc[1])
```

```python
print(df.loc[0:1, ["Name","Age"]])
```

## iloc (Integer Position-Based Indexing)

```python
print(df.iloc[1])
```

```python
print(df.iloc[0:2,0:2])
```
